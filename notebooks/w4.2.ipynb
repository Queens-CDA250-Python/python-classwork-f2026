```python
# Lord of Python
from typing import TypedDict, List, Literal, Dict
import random, time

class CharacterTypes(TypedDict):
    name: str
    health: int
    attack_power: int
    defense: int
    strength: str

strengths_weaknesses = {
    "fire": "water",
    "water": "earth",
    "earth": "fire",
}

total_points_available = 20

character: CharacterTypes = {
    "name": "Hero",
    "health": 90,
    "attack_power": 10,
    "defense": 10,
    "strength": "",
}

enemy_list: List[CharacterTypes] = [
    {
        "name": "Goblin",
        "health": 20,
        "attack_power": 5,
        "defense": 1,
        "strength": "water"
    }
]

locations = ["Forest"]

```

```python

for key in character:
    if key == "name":
        name = input("What is the name of your hero?")
        character["name"] = name
        character["name"]
    elif key == "strength":
        strength = input(f"What strength do you choose: {", ".join(strengths_weaknesses.keys())}")
        if strength.lower() not in [k.lower() for k in strengths_weaknesses]:
            print(f"{strength} is not a possible option, Bozo!")
            continue
        character["strength"] = strength.lower()
    else: 
        while True:
            int_value = input(f"You have {total_points_available} pts. How much do you want to add to {character["name"]}'s {key}")
            if not int_value.isnumeric():
                print("Hey, pass in an int, buddy!")
                continue
            int_value = int(int_value)
            if int_value > total_points_available:
                print(f"You do not have {int_value} pts available. You have {total_points_available} pts")
                continue
            total_points_available -= int_value # tpa = tpa - int_value
            character[key] += int_value
            print(f"You used {int_value} points.")
            print(f"Congratulations, {character["name"]} has {character[key]} {key} pts!")
            break
```
