```python
# Chess with Lists

pawns = ["P"] * 8
cool = ["R", "N", "B", "Q", "K", "B", "N", "R"]
empty = [" "] * 8

chess = [
    cool.copy(),
    pawns.copy(),
    empty.copy(),
    empty.copy(),
    empty.copy(),
    empty.copy(),
    pawns.copy(),
    cool.copy(),
]
chess
```

```python
chess[0][1] = " "
chess[2][2] = "N"
chess
```

```python
# tuple

# Not mutable (no changing),
# allows duplicates
# indexed
tpl = (1,2,2,3,"something")
tpl2 = tuple((1,2,3,4,5))
print(tpl)
print(tpl2)
```

```python
tpl.count(2)
tpl.index(2)
```

```python
# set
# can change (mutable)
# no duplicates
# unordered

set1 = {1,2,1,2,1,2,3} # the bad way
set2 = set([1,2,3,3,3,4,4,4,4]) # the good way
set2
```

```python
len(set2)
```

```python
set2.add(7)
len(set2)
```

```python
set2.remove(3)
len(set2)
```

```python
set2.difference(set1)
```

```python
set2 - set1
```

```python
# dictionary
# key: value pairs
# indexed by its key (not by its position)
# can change (mutable)
# keys cannot duplicate, values can.
edwin = {
    "name": "Edwin",
    "age": 12,
    "city": "Charlotte"
}
preston = {
    "name": "Preston",
    "city": "Charlotte",
    "age": 8,
}

person = {
    "name": "Nate",
    "age": 40,
    "city": "Charlotte",
    "children": [
        edwin,
        preston,
    ],
}
person
```

```python
person["age"]
```

```python
person["children"]
```

```python
person["children"][0]
```

```python
person["children"][0]["name"]
```
