```python
import pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from scipy import stats
from statsmodels.stats.multicomp import pairwise_tukeyhsd

url = "https://raw.githubusercontent.com/nkmwicz/worldcup2018data/refs/heads/main/cleaned_events_world_cup2018.csv"
# https://raw.githubusercontent.com/nkmwicz/worldcup2018data/refs/
# heads/main/cleaned_events_world_cup2018.csv
df = pd.read_csv(url)
df.head(1)
```

```python
df['shots'] = (df['eventName'] == "Shot").astype(int)
df['shots'].isna().sum()
```

```python
# get outcome
df['matchId'].unique()
```

```python
def get_outcome(row):
    match = row['matchId']
    event_team = row['teamId']
    teams, scores = match.split(",")
    t1,t2 = teams.split("-")
    s1,s2 = scores.split("-")
    s2 = s2.split("(")[0]
    s1 = int(s1)
    s2 = int(s2)
    if s1 > s2:
        if t1.strip() == event_team.strip():
            return "win"
        elif t2.strip() == event_team.strip():
            return "loss"
    elif s2 > s1:
        if t2.strip() == event_team.strip():
            return "win"
        elif t1.strip() == event_team.strip():
            return "loss"
    else: 
        return "tie"

df['outcome'] = df.apply(get_outcome, axis=1)
df['outcome'].unique()
```
