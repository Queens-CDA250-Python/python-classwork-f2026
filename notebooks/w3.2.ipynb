```python
from datetime import datetime
from typing import TypedDict
# comparisons
# == , != , < , > , <= , >=
class Person(TypedDict):
    name: str
    birth_year: int

drinking_age: int = 21
current_year: int = datetime.now().year

name: str = input("Enter your name: ")
birth_year: str = input("Enter your age: ")

person1: Person = {
    "name": name,
    "birth_year": birth_year,
}



if birth_year.isnumeric():
    person1['birth_year'] = int(birth_year)
    
    print(f"Is {person1['name']} old enough to drink? {(current_year - person1['birth_year']) >= drinking_age}")
else:
    print("A year is an integer, dumby!")
```
